```python
# Tokenization tools
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from string import punctuation
import re

# Scikit-Learn
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import CountVectorizer

# Coherence
import gensim.corpora as corpora
from gensim.models.coherencemodel import CoherenceModel
from gensim.utils import tokenize
```

```python
# class to calculate coherence score
class Metrics():

  def bert_coherence(self, sentences, topics, BERTopic_model):

    # Preprocess Documents
    documents = pd.DataFrame({
        "Document": sentences,
        "ID": range(len(sentences)),
        "Topic": topics}
    )
    documents_per_topic = documents.groupby(['Topic'], as_index=False).agg({'Document': ' '.join})
    cleaned_docs = BERTopic_model._preprocess_text(documents_per_topic.Document.values)

    # Extract vectorizer and analyzer from BERTopic
    vectorizer = BERTopic_model.vectorizer_model
    analyzer = vectorizer.build_analyzer()

    # Extract features for Topic Coherence evaluation
    words = vectorizer.get_feature_names_out()
    tokens = [analyzer(doc) for doc in cleaned_docs]
    dictionary = corpora.Dictionary(tokens)
    corpus = [dictionary.doc2bow(token) for token in tokens]
    topic_words = []
    for topic in set(topics):
        try:
            topic_data = BERTopic_model.get_topic(topic)
            if topic_data:  # Ensure the topic is valid
                topic_words.append([word for word, _ in topic_data])
        except Exception as e:
            print(f"Error processing topic {topic}: {e}")

    # Evaluate
    coherence_model = CoherenceModel(
        topics=topic_words,
        texts=tokens,
        corpus=corpus,
        dictionary=dictionary,
        coherence='c_v'
    )
    coherence = coherence_model.get_coherence()
    return coherence

  def t2v_coherence(self, sentences, topics):
      tokenized = [list(tokenize(doc)) for doc in sentences]
      id2word = corpora.Dictionary(tokenized)
      corpus = [id2word.doc2bow(text) for text in tokenized]

      # make sure you grab the topic words from the topic model and convert them to a list
      coherence_model = CoherenceModel(
          topics = topics,
          texts=tokenized,
          corpus=corpus,
          dictionary=id2word,
          coherence='c_v',
          topn=10
      )
      coherence = coherence_model.get_coherence()
      return coherence
```

```python
# preprocess the text
# remove hastags, mentions and links
# Define a function to clean text by removing hashtags, mentions, and links
def clean_text(text):
    # Remove hashtags, mentions, and links using regex
    text = re.sub(r'@\w+', '', text)  # Remove mentions
    text = re.sub(r'#\w+', '', text)  # Remove hashtags
    text = re.sub(r'http\S+|www.\S+', '', text)  # Remove links
    # Remove extra spaces and newlines
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
##Functions for creating word clouds
def create_wordcloud(model, topic_num):
  words = model.get_topic(topic_num)
  word_freq = {word: freq for word, freq in words}
  wc = WordCloud(background_color="white", max_words=1000).generate_from_frequencies(word_freq)
  # Create a figure and axis
  fig, ax = plt.subplots(figsize=(8, 10))
  ax.imshow(wc, interpolation='bilinear')
  ax.axis('off')  # Remove axis
  return fig
```

```python
# Import required libraries
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
```

```python
# Load the IndicBERT model and tokenizer for sentiment analysis
model_name = "AI4Bharat/indic-bert"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=3)
```

```python
# Define a function to perform sentiment analysis
def analyze_sentiment(replies):
    sentiments = []
    for text in replies:
        # Tokenize and prepare input
        inputs = tokenizer(text, return_tensors='pt', truncation=True, padding=True, max_length=512)

        # Run the model and get outputs
        with torch.no_grad():
            outputs = model(**inputs)

        # Extract sentiment score
        scores = torch.softmax(outputs.logits, dim=1).numpy()[0]
        sentiment = scores.argmax()  # Get the index of the highest score

        # Map sentiment index to labels
        if sentiment == 0:
            sentiments.append('negative')
        elif sentiment == 1:
            sentiments.append('neutral')
        else:
            sentiments.append('positive')
    return sentiments
```

```python
# Apply sentiment analysis to each reply in the grouped DataFrame
merged_df['sentiment'] = merged_df['replies'].apply(analyze_sentiment)
```

```python
# Function to get sentiment using ChatGPT API
def get_sentiment(tweet_context, reply):
    # Create a prompt combining tweet context and reply
    prompt = f"""
    Given a tweet and a list of replies, classify the sentiment of each reply as Positive, Neutral, or Negative based on the context of the tweet in one word.
    Tweet Context: "{tweet_context}"
    Reply: "{reply}"
    Return the sentiments as a list like this example: ["Negative", "Negative", "Positive", ...]
    """

    # Make API request
    completion = client.chat.completions.create(
    model="gpt-4o",
    store=True,
    messages=[
        {"role": "system", "content": "You are a helpful assistant skilled in sentiment analysis."},
        {"role": "user", "content": prompt}
    ]
)
    # Extract the sentiment from the response
    sentiment = completion.choices[0].message.content
    return sentiment
```
